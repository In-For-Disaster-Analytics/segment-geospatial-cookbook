# Segment Geospatial (SAM 3) on Lonestar6

Interactive environment for `segment-geospatial` with the Meta **SAM 3** backend.
The conda environment `segment-geospatial` is pre-installed (see `.binder/environment.yaml`).

SAM 3 requires a CUDA GPU (use the **GPU** app) and Hugging Face checkpoint access
(`facebook/sam3`, `facebook/sam3.1`). Authenticate with `huggingface-cli login` (or set
the `HF_TOKEN` env var) and accept the license before using `SamGeo3`.

```python
import os
import leafmap
from samgeo import SamGeo3

# BPE tokenizer path — run.sh downloads it to $SCRATCH and patches samgeo3,
# but pass it explicitly as a safety net.
BPE_PATH = os.path.join(os.environ.get("HOME", "/tmp"), "bpe_simple_vocab_16e6.txt.gz")

print("segment-geospatial import OK")
```

```python
# Open-vocabulary text-prompt segmentation example.
# Point `image` at a GeoTIFF/imagery in the mounted working directory.
sam = SamGeo3(backend="meta", bpe_path=BPE_PATH)
sam.set_image("input.tif")
sam.generate_masks("tree")
# Input was a GeoTIFF, so masks save as a GeoTIFF. Use a .png path for non-geospatial input.
# SamGeo3 saves rasters (GeoTIFF/PNG) only -- to get vector polygons (GeoPackage),
# vectorize the mask raster afterwards (rasterio/geocube or leafmap.raster_to_vector).
sam.save_masks("masks.tif")
```
